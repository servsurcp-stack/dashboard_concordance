```python
cols = [
    'ANOMALIE DE CHARGEMENT',
    'ANOMALIE DE VEHICULE',
    'ANOMALIE SUIVI DE TOURNEE'
]

# (optionnel) traiter les chaînes vides comme NaN
df_agence[cols] = df_agence[cols].replace(r'^\s*$', pd.NA, regex=True)

# nombre d'anomalies par catégorie = nb de valeurs non nulles
compte_par_categorie = (
    df_agence[cols].notna().sum()
    .rename('nb_anomalies')
    .to_frame()
    .reset_index()
    .rename(columns={'index': 'categorie'})
)

print(compte_par_categorie)
```
